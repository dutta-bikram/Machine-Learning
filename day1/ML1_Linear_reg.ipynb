```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df = pd.read_csv('file.csv')
df.head(3)
```

```python
%matplotlib inline
plt.xlabel('area(sq.ft.)')
plt.ylabel('price(US$)')
plt.scatter(df.area,df.price, color='red', marker='+')
```

```python
reg = linear_model.LinearRegression()
reg.fit(df[['area']],df['price'])
```

```python
reg.predict(pd.DataFrame({'area':[5000]})).round(2)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
d = pd.read_csv('query.csv');
d.head(3)
```

```python
p = reg.predict(d).round(2)
d['prices'] = p;
d.to_csv('output.csv', index=False)
```

```python
%matplotlib inline
plt.xlabel('area(sq.ft.)')
plt.ylabel('price(US$)')
plt.scatter(df.area,df.price, color='red', marker='+')
plt.plot(df.area, reg.predict(df[['area']]), color='blue')
```
